Modificamos a tipo fecha las variables *timestamp* y *release_date*

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
```

```python
df["timestamp"].dtype
```

```python
df["timestamp"].head()
```

```python
df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce")
```

```python
df["release_date"].dtype
```

```python
df["release_year"] = df["release_date"].dt.year
df["rating_year"] = df["timestamp"].dt.year
```

```python
#cambiamos la columna "tmsbId" de tipo float a tipo int
df["tmdbId"] = df["tmdbId"].astype(int)
```

```python
id_columns = ["userId", "movieId", "imdbId", "tmdbId", "movieId", "id"]
```
